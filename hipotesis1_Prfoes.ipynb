```python
columnas = ["participant_id", "age", "education_level", "bp_systolic", "bp_diastolic", "blood_urea_nitrogen", "bicarbonate", "urine_creatinine", "diabetes_diagnosed", "insulin_use", "diabetes_pills", "ever_smoked","current_smoker", "ckd_present"]
data[columnas] = data[columnas].round().astype("Int64")
data.head()
```

```python
sin_diabetes = data["diabetes_diagnosed"] == 2
con_insulina = data["insulin_use"] == 1
con_pastillas = data["diabetes_pills"] == 1

print((sin_diabetes & con_insulina).sum())                    
print((sin_diabetes & con_pastillas).sum())                   
print((sin_diabetes & (con_insulina | con_pastillas)).sum())  
#las pildoras para la diabetes no son indicador de poseer la enfermedad, ya que puede tener pre diabetes y tomar la pastilla, por lo que indico que no tiene la enfermedad diagnosticada.
```

```python
#modificamos los valores 7 y 9 de las columnas donde este valor indica que no sabe o no quiere responder
import numpy as np
data["education_level"] = data["education_level"].replace(9, np.nan)
data["current_smoker"] = data["current_smoker"].replace(7, np.nan)
data["current_smoker"] = data["current_smoker"].replace(9, np.nan)
data["ever_smoked"] = data["ever_smoked"].replace(7, np.nan)
data["ever_smoked"] = data["ever_smoked"].replace(9, np.nan)
data["diabetes_diagnosed"] = data["diabetes_diagnosed"].replace(9, np.nan)
data["diabetes_pills"] = data["diabetes_pills"].replace(9, np.nan)
```

```python
#La hipotesis univariada: La cantidad de personas fumadoras  del dataset superan a la media mundial la cual se ubica en el 20% de la poblacion
# H0: p >= 0.20 (el porcentaje de fumadores actuales es igual o mayor a la media mundial)
# H1: p <  0.20 (el porcentaje de fumadores actuales es menor a la media mundial)
import scipy.stats as stats
from statsmodels.stats.proportion import proportions_ztest
alpha = 0.05
p0 = 0.20
poblacion_observada = data[data["ever_smoked"].notna()]#filtramos los menores a 18 y los que no respondieron bien la columna
n_total_observados = len(poblacion_observada)

total_fumadores_activos = data["current_smoker"].isin([1, 2]).sum()
no_fumadores_total = ((poblacion_observada["ever_smoked"] == 2) | (poblacion_observada["current_smoker"] == 3)).sum()

porcentaje_de_fumadores = (total_fumadores_activos / n_total_observados)*100

print(f"Total de población adulta observada: {n_total_observados}")
print(f"Total de fumadores activos: {total_fumadores_activos}")
print(f"Total de personas que no fuman actualmente: {no_fumadores_total}")
print(f"Porcentaje de fumadores activos: {porcentaje_de_fumadores:.2f}%")



#utilizamos testz porque este sirve para  comparar un porcentaje observado contra un valor fijo, en este caso el 20%
z_estadistico, p_valor = proportions_ztest(
    count=total_fumadores_activos, 
    nobs=n_total_observados, 
    value= 0.20, 
    alternative='smaller'  # Buscamos demostrar que es menor al 20%
)
print(f"Estadístico Z: {z_estadistico:.4f}")
print(f"Valor p: {p_valor:.4f}")

if p_valor < alpha:
    print("Se RECHAZA la hipótesis nula (H0). El porcentaje de fumadores activos es significativamente menor al 20%.")
else:
    print("NO se rechaza la hipótesis nula (H0). No hay evidencia suficiente para afirmar que es menor al 20%.")
#AGREGAR EL GRAFICO
# Como pudimos observar, el p-valor es 0.0000 (menor al alpha = 0.05), 
# lo que indica que rechazamos la hipótesis nula (H0). 
# Esto nos permite concluir que el porcentaje de fumadores activos en nuestro dataset (14.65%,) 
# es significativamente menor al 20%, validando que se encuentra por debajo de la media mundial.
```
